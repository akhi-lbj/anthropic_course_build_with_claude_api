```python
# Create an API client
from anthropic import Anthropic

client  = Anthropic()
model = "claude-haiku-4-5"
```

```python
from email import message


def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)


def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=messages 
    )
    return message.content[0].text
```

```python
# Initialize a list to store conversation messages
messages = []

# Add a user message asking for a one-sentence explanation of quantum computing
add_user_message(messages, "Determine quantum computing in one sentence")

answer = chat(messages)

add_assistant_message(messages, answer)

add_user_message(messages, "Write another sentence")

answer_1=chat(messages)
answer_1


```
