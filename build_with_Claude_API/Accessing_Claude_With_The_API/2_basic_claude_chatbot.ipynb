A simple multi-turn chatbot using the Anthropic API.
Run the cells in order, then use the last cell to chat.

```python
import anthropic

# ── Your API key ────────────────────────────────────────────────
from dotenv import load_dotenv
load_dotenv()
```

```python
from anthropic import Anthropic

client  = Anthropic()
```

```python
# ── Helper 1: Add a message to the conversation list ────────────
def add_message(messages, role, content):
    """Append a {role, content} dict to the messages list."""
    messages.append({"role": role, "content": content})
    return messages


# ── Helper 2: Call the Claude API ───────────────────────────────
def call_claude(messages, model="claude-haiku-4-5", max_tokens=1024):
    """Send the full conversation history and return Claude's reply text."""
    response = client.messages.create(
        model=model,
        max_tokens=max_tokens,
        messages=messages
    )
    return response.content[0].text


# ── Helper 3: Print the assistant reply nicely ──────────────────
def print_reply(text):
    """Pretty-print Claude's response."""
    print(f"\nClaude: {text}\n")
    print("-" * 60)
```

```python
# ── Chatbot loop ────────────────────────────────────────────────
# Type  'quit'  or  'exit'  to end the conversation.

messages = []   # shared conversation history

print("Claude Chatbot — type 'quit' to exit\n" + "=" * 60)

while True:
    # 1. Prompt the user for input
    user_text = input("You: ").strip()

    if not user_text:
        continue
    if user_text.lower() in ("quit", "exit"):
        print("Goodbye!")
        break

    # 2. Add the user message to the list
    add_message(messages, "user", user_text)

    # 3. Call the Claude API
    reply = call_claude(messages)

    # 4. Add Claude's reply to the list
    add_message(messages, "assistant", reply)

    # 5. Print the generated text
    print_reply(reply)

    # 6. Loop repeats automatically
```
